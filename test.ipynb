```python
import numpy as np
from PIL import Image

def cambiar_negro_a_blanco_con_tolerancia(ruta_entrada, ruta_salida, tolerancia=30):
    """
    Cambia los píxeles oscuros (cercanos al negro) a blanco, conservando la transparencia.
    
    :param tolerancia: Valor de 0 a 255. Define qué tan "oscuro" debe ser un píxel 
                       para ser considerado negro. Un valor de 30 a 50 suele funcionar bien.
    """
    # 1. Abrir la imagen y asegurar que esté en modo RGBA
    imagen = Image.open(ruta_entrada).convert("RGBA")
    
    # 2. Convertir a array de NumPy
    datos = np.array(imagen)
    
    # 3. Separar los canales de color
    rojo = datos[:, :, 0]
    verde = datos[:, :, 1]
    azul = datos[:, :, 2]
    
    # 4. Crear una máscara usando la tolerancia
    # Identifica los píxeles donde TODOS los canales (R, G, B) sean menores que la tolerancia
    pixeles_oscuros = (rojo < tolerancia) & (verde < tolerancia) & (azul < tolerancia)
    
    # 5. Cambiar el color de esos píxeles a blanco (255, 255, 255)
    datos[:, :, :3][pixeles_oscuros] = [255, 255, 255]
    
    # 6. Guardar la imagen
    imagen_resultado = Image.fromarray(datos)
    imagen_resultado.save(ruta_salida)
    
    print(f"Imagen procesada (con tolerancia {tolerancia}) guardada en: {ruta_salida}")

# --- Ejecución del código ---
if __name__ == "__main__":
    imagen_original = "mi_imagen.png" 
    imagen_modificada = "mi_imagen_blanca.png"
    
    # Puedes ajustar el número al final. 
    # Si aún quedan partes oscuras, súbelo (ej. a 50 o 60).
    # Si se están volviendo blancos colores que no querías (como azul oscuro), bájalo (ej. a 15).
    cambiar_negro_a_blanco_con_tolerancia(imagen_original, imagen_modificada, tolerancia=40)
```
